The code below tests every possible positive whole-number width for this problem.

**My prediction for the width-5 line:**  

Run the code when your prediction is recorded. Please do not modify this code yet.

````python
perimeter = 24

for width in range(1, 12):
    length = perimeter / 2 - width
    area = width * length
    print("Width:", width, "Length:", length, "Area:", area)
````
