## 4 · Make a table of differences (12 minutes)
Run this code. It prints the term number, sequence value, and the amount added **after** printing.

```python
term = 1
increase = 2
print("n | term | next increase")
for n in range(1, 11):
    print(n, "|", term, "|", increase)
    term = term + increase
    increase = increase + 1
```
