## 2. Generate ordered pairs (8–18 min)
Run the cell. Notice that `zip(xs, ys)` pairs each x-value with its matching y-value.


```python
xs = list(range(-3, 4))
ys = [2*x + 1 for x in xs]
for x, y in zip(xs, ys):
    print(f"({x}, {y})")
```

## 3. Plot the points (18–33 min)
Run this code to see the ordered pairs on a coordinate plane. The line connects the points to show the relationship.


```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
plt.plot(xs, ys, 'o-', label='y = 2x + 1')
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)
plt.xticks(xs)
plt.grid(alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('From a rule to a graph')
plt.legend()
plt.show()
```

## 4. Investigate slope (33–45 min)
Compare the change in y with the change in x between any two distinct points.

\[\text{slope} = \frac{\text{change in }y}{\text{change in }x}\]

**TRY:** Use points (−1, −1) and (2, 5). Calculate the slope. Then choose a different pair of points and check again.

**Your work:**



```python
p1 = (-1, -1)
p2 = (2, 5)
slope = (p2[1] - p1[1]) / (p2[0] - p1[0])
print('Slope:', slope)
```

## 5. Test a conjecture (45–53 min)
What happens if we keep the same coefficient of x but change the constant? Compare **y = 2x + 1** and **y = 2x − 3**.

```python
y1 = [2*x + 1 for x in xs]
y2 = [2*x - 3 for x in xs]
plt.figure(figsize=(7,5))
plt.plot(xs, y1, 'o-', label='y = 2x + 1')
plt.plot(xs, y2, 's-', label='y = 2x - 3')
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)
plt.grid(alpha=0.3)
plt.xlabel('x'); plt.ylabel('y'); plt.legend(); plt.show()
```
